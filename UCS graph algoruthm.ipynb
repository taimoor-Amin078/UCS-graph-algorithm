```python
import heapq  # Import the heapq module to use a priority queue (min-heap)

def uniform_cost_search(graph, start, goal):
    # Initialize the Priority Queue with a tuple: (cumulative_cost, current_node, path_list)
    # The cost is first because heapq sorts by the first element of the tuple.
    priority_queue = [(0, start, [start])]

    # Create a dictionary to keep track of the minimum cost to reach each visited node.
    visited = {}

    # Continue the search as long as there are nodes in the priority queue.
    while priority_queue:
        # Pop the node with the lowest cumulative cost (this is the 'Uniform' part).
        (cost, current_node, path) = heapq.heappop(priority_queue)

        # Check if we have reached the destination.
        # In UCS, the first time we pop the goal, it is guaranteed to be the cheapest path.
        if current_node == goal:
            return path, cost

        # Optimization: If we have already reached this node with a lower or equal cost,
        # there is no point in exploring its neighbors again.
        if current_node in visited and visited[current_node] <= cost:
            continue

        # Mark the current node as visited and store the cost it took to get here.
        visited[current_node] = cost

        # Look at all neighbors of the current node defined in our graph.
        # .get() handles nodes that might not have any outgoing edges (like node 'G').
        for neighbor, weight in graph.get(current_node, {}).items():
            # Calculate the total cost from the 'start' node to this neighbor.
            new_cost = cost + weight

            # Push the updated cost, the neighbor's name, and the updated path into the heap.
            # The heap will automatically re-sort so the cheapest path is always on top.
            heapq.heappush(priority_queue, (new_cost, neighbor, path + [neighbor]))

    # If the queue becomes empty and the goal was never reached, return None.
    return None, float('inf')

# --- Defining the search graph ---
# Each key is a node, and its value is a dictionary of {neighbor: edge_cost}.
graph = {
    'S': {'A': 3, 'B': 2, 'C': 7},
    'A': {'D': 3, 'E': 8, 'G': 15},
    'B': {'G': 20},
    'C': {'G': 6},
    'D': {},
    'E': {},
    'G': {}
}

# Call the function with 'S' as start and 'G' as goal.
path, total_cost = uniform_cost_search(graph, 'S', 'G')

# Output the results.
if path:
    print(f"Optimal Path: {' -> '.join(path)}")
    print(f"Minimum Cost: {total_cost}")
else:
    print("No path found.")
```
